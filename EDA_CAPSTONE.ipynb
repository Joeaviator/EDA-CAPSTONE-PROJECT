```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
from google.colab import files
file = files.upload()
```

```python
df = pd.read_csv('Titanic Dataset.csv')
df.head()
```

```python
df.info()
```

```python
df=df.drop(columns=['Name', 'Cabin', 'Ticket', 'PassengerId'], axis=1)
```

```python
df.head()
```

```python
#Statistical Distribuion
df.describe()
```

```python
sns.boxplot(df['Survived'])
plt.show()
```

```python
sns.boxplot(df['Age'])
plt.show()
```

```python
sns.countplot(x='Survived',data=df, palette='rainbow')
plt.title("Passenger Survival Distribution ")
plt.show()
```

```python
sns.countplot(x='Gender',data=df, palette='turbo')
plt.title("Passenger Gender Distribution ")
plt.show()
```

```python
sns.countplot(x='Gender', hue='Survived', data=df, palette='turbo')
plt.show()
```

```python
sns.barplot(x='Gender', y='Survived', data=df, palette='rainbow')
plt.show()
```

```python
sns.histplot(x='Age', hue='Survived', data=df, palette='turbo')
```

```python
binned_age=pd.cut(df['Age'], bins=[0,20,30,40,50,60,70,80],labels = ['Babies', 'Children', 'Youth', 'Young Adult', 'Adult', 'Middle Age', 'Senior'])
sns.barplot(x=binned_age, y='Survived', data=df, palette='turbo')
```

```python
sns.countplot(x='SibSp',data=df, palette='rainbow')
plt.show()
```

```python
sns.countplot(x='Pclass',data=df, palette='rainbow')
plt.show()
```

```python
binned_fare=pd.cut(df['Fare'], bins=[0,50,100,180,250],labels=['Low', 'Medium', 'High', 'Very High'])
sns.countplot(x=binned_fare, hue='Pclass', data=df, palette='turbo')
```

```python
sns.pairplot(df)
```
